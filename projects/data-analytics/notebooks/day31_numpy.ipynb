```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
```

```python
print(np.arange(0, 10, 2))
```

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
], dtype=np.int32)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

```python
arr = np.arange(12)
print(arr.reshape(3, 4))
```

```python
# vectorization
numbers = np.array([1, 2, 3, 4, 5])
result = numbers * 10
print(result)
```

```python
predicted = np.array([80, 90, 70, 60])
actual = np.array([85, 88, 75, 65])
errors = predicted - actual
absolute_errors = np.abs(errors)
mae = np.mean(absolute_errors)

print(mae)
```

```python
# Exercise 6
predicted = np.array([100, 200, 300, 400])
actual = np.array([110, 190, 320, 380])
error = predicted - actual
absolute_errors = np.abs(error)
mae = np.mean(absolute_errors)
print("Error: ", error)
print("Absolute Error: ", absolute_errors)
print("Mean Absolute Error: ", mae)
```
