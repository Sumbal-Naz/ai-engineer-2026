```python
import pandas as pd
```

```python
# Exercise 1: Create a Series
numbers = pd.Series([10, 20, 30, 40, 50])
print(numbers)
print(numbers.values)
print(numbers.index)
print(numbers.dtype)
```

```python
# Exercise 2: Create a DataFrame
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Bilal"],
    "Age": [25, 28, 30, 26, 35],
    "Salary": [60000, 75000, 90000, 65000, 100000]
}
salaries = pd.DataFrame(data)
print(salaries.shape)
print(salaries.columns)
print(salaries.index)
print(salaries.dtypes)
```

```python
# Exercise 3: Inspect
print(salaries.head(3))
print(salaries.tail(2))
salaries.info()
print(salaries.describe())
```

```python
# Exercise 4 Select
print(salaries["Name"])
print(salaries[["Name", "Salary"]])
print(salaries.loc[0])
print(salaries.iloc[0])
print("Sara Salary :" , salaries.loc[salaries["Name"] == "Sara", "Salary"])
```

```python
# Exercise # 5 Filtering
emp_28 = salaries[salaries["Age"] > 28]
print("Employees older than 28")
print(emp_28["Name"])

emp_70 = salaries[salaries["Salary"] > 70000]
print("Employees earning more than 70,000")
print(emp_70["Name"])

emp_28_70 = salaries[(salaries["Age"] > 28) & (salaries["Salary"] > 70000)]
print("Employees older than 25 AND earning more than 70,000")
print(emp_28_70["Name"])

emp_28_90 = salaries[(salaries["Age"] > 28) | (salaries["Salary"] > 90000)]
print("Employees younger than 28 OR earning more than 90,000")
print(emp_28_90["Name"])
```

```python
# Exercise 6 Statistics

print("Average Age: ", salaries["Age"].mean())
print("Average Salary: ", salaries["Salary"].mean())
print("Maximum Salary: ", salaries["Salary"].max())
print("Minimum Salary: ", salaries["Salary"].min())
print("Total Salary: ", salaries["Salary"].sum())
print("Salary Standard Deviation: ", salaries["Salary"].std())
```
