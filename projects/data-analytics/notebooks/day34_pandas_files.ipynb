```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Bilal"],
    "Age": [25, 28, 30, 26, 35],
    "Salary": [60000, 75000, 90000, 65000, 100000],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
}

df = pd.DataFrame(data)
```

```python
# Exercise 1: CSV
df.to_csv(
    "../data/raw/employees.csv",
    index=False
)

employee_data = pd.read_csv("../data/raw/employees.csv")
print(employee_data)
print(employee_data.shape)
print(employee_data.columns)
print(employee_data.dtypes)
```

```python
# Exercise 2: Excel
df.to_excel(
    "../data/raw/employees.xlsx",
    index=False
)

employee_data_excel = pd.read_excel("../data/raw/employees.xlsx")
print(employee_data_excel)
print(employee_data_excel.shape)
print(employee_data_excel.columns)
```

```python
# Exercise 3: JSON
df.to_json(
    "../data/raw/employees.json",
    orient="records"
)
employee_data_json = pd.read_json(
    "../data/raw/employees.json"
)
print(employee_data_json)
print(employee_data_json.shape)
print(employee_data_json.dtypes)
```

```python
# Exercise 4: Comparing Formats

print(employee_data.equals(employee_data_excel))
print(employee_data.equals(employee_data_json))
print(employee_data_json.equals(employee_data_excel))
```
