```python
import pandas as pd
import numpy as np

data = {
    "Name": [
        "Ali Khan",
        "Sara Ahmed",
        "Ahmed Raza",
        "Ayesha Malik",
        "Bilal Shah"
    ],
    "Department": [
        "IT",
        "HR",
        "IT",
        "Finance",
        "IT"
    ],
    "Email": [
        "ali.khan@gmail.com",
        "sara.ahmed@yahoo.com",
        "ahmed.raza@gmail.com",
        "ayesha.malik@company.com",
        "bilal.shah@gmail.com"
    ],
    "JoiningDate": [
        "2024-01-15",
        "2023-08-20",
        "2025-03-10",
        "2022-06-05",
        "2024-11-25"
    ],
    "Salary": [
        60000,
        75000,
        90000,
        65000,
        80000
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Exercise 1 — Basic string transformations

df["NameLower"] = df["Name"].str.lower()
df["NameUpper"] = df["Name"].str.upper()
df["FirstName"] = df["Name"].str.split().str[0]
df["LastName"] = df["Name"].str.split().str[-1]
df["NameLength"] = df["Name"].str.len()
```

```python
# Exercise 2 — Email features

df["IsGmail"] = df["Email"].str.contains("@gmail.com")
df["EmailDomain"] = df["Email"].str.split("@").str[-1]
```

```python
#Exercise 3 — Date features

df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
df["JoiningYear"] = df["JoiningDate"].dt.year
df["JoiningMonth"] = df["JoiningDate"].dt.month
df["JoiningDay"] = df["JoiningDate"].dt.day
df["JoiningWeekday"] = df["JoiningDate"].dt.day_name()
```

```python
# Exercise 4: Tenure

today = pd.Timestamp("2026-10-07")

df["TenureDays"] = (today - df["JoiningDate"]).dt.days
df["TenureYears"] = df["TenureDays"] / 365
df["TenureYears"] = df["TenureYears"].round(1)
```

```python
# Exercise 5: Experience classification

df["ExperienceLevel"] = np.where(
    df["TenureYears"] < 2,
    "Junior",
    "Experienced"
)
print(df[["Name", "Salary", "ExperienceLevel"]])
```

```python
# Exercise 6: Salary Classification

df["SalaryLevel"] = np.select(
    [
        df["Salary"] < 70000,
        df["Salary"] <= 80000
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)
print(df[["Name", "Salary", "SalaryLevel"]])
```
